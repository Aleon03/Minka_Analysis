Script para realizar la descarga de los siguiente parametros y almacenarlos en un excell

* id_obs 
* user_id
* observed_on
* created_at
* taxon_id
* taxon_name
* latitude, longitude

```python
import pandas as pd
import requests
```

```python
API_PATH = 'https://api.minka-sdg.org/v1/'
```

```python
def get_data(project_id: int, filename: str):
    url = f'{API_PATH}observations?project_id={project_id}'
    per_page = 200
    total_number = requests.get(url).json()['total_results']
    data = []

    for i in range(1,(total_number // per_page) + 2):
        results = requests.get(url).json()['results']
        for i in range(len(results)):
            if results[i]['quality_grade'] == 'research':
                id_obs = results[i]['id']
                user_id = results[i]['user']['id']
                username = results[i]['user']['login']
                observed_on = results[i]['observed_on_string']
                created_at = results[i]['created_at_details']['date']
                taxon_id = results[i]['taxon']['id']
                taxon_name = results[i]['taxon']['name']
                longitud = results[i]['geojson']['coordinates'][0]
                latitud = results[i]['geojson']['coordinates'][1]
                minka_page = results[i]['uri']
                species_guess = results[i]['species_guess']
                place_guess = results[i]['place_guess']
            
                data.append([species_guess, id_obs, taxon_name, taxon_id,
                         observed_on, created_at, username, user_id, 
                          longitud, latitud, place_guess, minka_page])
        else:
            continue
    
    df = pd.DataFrame(data, columns = ['Nombre de la observacion', 'ID de la observacion', 'Nombre cientifico', 'ID del taxon',
                                        'Fecha de la observacion', 'Fecha de creacion', 'Nombre de usuario', 'ID de usuario',
                                        'Longitud', 'Latitud', 'Lugar de la observacion', 'Link a la página de Minka'])
    df.to_csv(filename, index = False)
    
    return df
```

```python
get_data(264, 'data.csv')
```
