```python
API_PATH = 'https://api.minka-sdg.org/v1'
```

```python
def get_main_metrics(project_id: int):
    
    observations = f'{API_PATH}observations?'
    url_observations = f'{observations}&project_id={project_id}&quality_grade=research'
    total_observations = requests.get(url_observations).json()['total_results']

    species = f'{API_PATH}observations/species_counts?'
    url_species = f'{species}&project_id={project_id}'
    total_species = requests.get(url_species).json()["total_results"]

    observers = f'{API_PATH}observations/observers?'
    url_observers = f"{observers}&project_id={project_id}"
    total_participants = requests.get(url_observers).json()["total_results"]

    identifiers = f'{API_PATH}observations/identifiers?'
    url_identifiers = f"{identifiers}&project_id={project_id}"
    total_identifiers = requests.get(url_identifiers).json()["total_results"]

    return total_observations, total_species, total_participants, total_identifiers
```

```python
def get_daily_metrics(project_id: int) -> pd.DataFrame:
    
    today = datetime.datetime.today()
    metrics = {'dates': [], 'observations': [], 'species': [], 'observers': [], 'identifiers': []}

    for i in range(30):

        date = (today - datetime.timedelta(days=i)).strftime('%Y-%m-%d')
        metrics['dates'].append(date)

        observations = f'{API_PATH}observations?'
        url_observations_daily = f'{observations}&project_id={project_id}&d1={date}&d2={date}'
        metrics['observations'].append(requests.get(url_observations_daily).json()['total_results'])

        species = f'{API_PATH}observations/species_counts?'
        url_species_daily = f'{species}&project_id={project_id}&d1={date}&d2={date}'
        metrics['species'].append(requests.get(url_species_daily).json()["total_results"])

        observers = f'{API_PATH}observations/observers?'
        url_observers_daily = f"{observers}&project_id={project_id}&d1={date}&d2={date}"
        metrics['observers'].append(requests.get(url_observers_daily).json()["total_results"])

        identifiers = f'{API_PATH}observations/identifiers?'
        url_identifiers_daily = f"{identifiers}&project_id={project_id}&d1={date}&d2={date}"
        metrics['identifiers'].append(requests.get(url_identifiers_daily).json()["total_results"])

    return pd.DataFrame(metrics)
```

```python
def get_heatmap_obs(project_id: int):
    
    observations = f'{API_PATH}observations?'
    url_observations = f'{observations}&project_id={project_id}&quality_grade=research'
    per_page = 200
    total_number = requests.get(url_observations).json()['total_results']

    latitudes = []
    longitudes = []

    # Obtener datos paginados
    for page in range(1, (total_number // per_page) + 2):
        paginated_url = f"{url_observations}&per_page={per_page}&page={page}"
        results = requests.get(paginated_url).json().get('results', [])

        for obs in results:
                try:
                    latitudes.append(float(obs['geojson']['coordinates'][1]))
                    longitudes.append(float(obs['geojson']['coordinates'][0]))
                except (KeyError, TypeError, ValueError):
                    continue  # Ignorar errores en coordenadas

    # Verificar que tenemos datos
    if not latitudes or not longitudes:
        print("No se encontraron observaciones válidas.")
        return None

    # Crear mapa centrado en la media de las coordenadas
    m = folium.Map(location=[np.mean(latitudes), np.mean(longitudes)], zoom_start=10)

    # Crear el HeatMap y agregarlo al mapa
    heat_data = list(zip(latitudes, longitudes))
    HeatMap(heat_data).add_to(m)

    # Guardar y mostrar el mapa
    m.save('heatmap.png')
    return m


```
