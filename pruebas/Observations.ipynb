## Script con las funciones más útiles e importantes de la funcion de la API de /observations.

##### En la siguiente url se obtienen todos los resultados totales de las observaciones realizadas en la web MINKA

```python
# Establecemos una variable global para no tener que escribir esta parte siempre en las URLs, para evitar errores al escribirlo
API_PATH = "https://api.minka-sdg.org/v1"
```

```python
import requests
import pandas as pd

url = f"{API_PATH}/observations" # f delante de las comillas del string nos permite incluir la variable dentro (entre llaves)

print(requests.get(url).json()["total_results"]) # Se obtiene el número total de resultados
```

```python
url = f"{API_PATH}/observations"

'''En results no entiendo el orden de los resultados, ya que esta ordenado de diferente forma que el resultado en Firefox. Donde si sale ordenado
correctamente en orden cronologico de publicación más reciente '''

results = requests.get(url).json()["results"]
```

NOTA: El orden es cronológico, mira si iteramos por el campo "created_at":

```python
for result in results:
    print(f"ID {result['id']}: fecha de subida {result['created_at']}, fecha de observación {result['observed_on']}")
```

La siguiente url muestra las observaciones realizadas por un usuario donde se muestran 100 resultados por pagina

```python

url = f"{API_PATH}/observations?user_login=xasalva&per_page=100"

'''En esta url obtenemos el numero de observaciones que se han hecho en la plataforma de minka-sdg,
por el usuario xasalva, y se muestran 100 observaciones por pagina. IMPORTANTE'''

requests.get(url).json()["results"]
```

La siguiente url, nos muestra los resultados de una observación en concreto. 

Como se indica en los comentarios, seria interesante que el usuario pudiese acceder a cualquier ID que le interesase y de tal forma obtener la información del organismo.

NOTA: Esto lo hablamos, el objetivo del análisis es aportar información a los gestores de proyectos. Es decir, datos agregados.

```python
obs_id = 417841
url = f"{API_PATH}/v1/observations/{obs_id}" # para que quede claro la parte que es modificable, la metemos como variable

'''De esta forma se accede a una ID de una observación especifica. Pensando en la finalidad del proyecto y en la obtención de un paper
lo mejor seria que el usuario pudiera acceder a la ID de la observación que le interese, y que se le muestre la información de la observación'''

requests.get(url).json()['results'] # El resultado da lo que quiero pero las imagenes no corresponden a la observación sino a la imagen que se aportan en los comentarios
```

Se muestran las observaciones totales de un usuario. Para el futuro objetivo del proyecto, seria interesante que se pudiesen identificar el numero de veces que se ha observado un ID (organismos) en los diferentes meses.

NOTA: Esa búsqueda la puedes hacer por el "taxon_id", el ID de la taxonomía que quieres revisar en lugar de usar el usuario. El taxon_id de una especie lo puedes encontrar aquí: [https://minka-sdg.org/taxa/](https://minka-sdg.org/taxa/)

Por ejemplo, para Coris julis: https://minka-sdg.org/taxa/34835-Coris-julis

El taxon_id es 34835, el número que aparece en la URL.

```python
user_name = "xasalva"
url = f"{API_PATH}/observations/histogram?user_login={user_name}&date_field=observed&interval=month_of_year"

''' La siguiente url, nos enseña la cantidad de datos observados por el usuario xasalva, en un intervalo de tiempo de un mes.
Podria ser interesante para registrar la cantidad de datos que se han observado en un mes y poder identificar las especies observadas en ese mes'''

month_of_year = requests.get(url).json()['results']["month_of_year"]

total = sum(month_of_year.values())

print(requests.get(url).json())

print(f" Numero de observaciones totales del usuario: {total}")
```
