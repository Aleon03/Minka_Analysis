Graficos 

```python
import requests
import seaborn as sns
import matplotlib.pyplot as plt
import folium
from folium.plugins import HeatMap
import numpy as np
import pandas as pd
```

```python
API_PATH = 'https://api.minka-sdg.org/v1/'
```

GRAFICOS OBSERVACIONES

Histograma

```python
def get_hist_obs(project_id: int):
    url  = f'{API_PATH}observations/histogram?project_id={project_id}&date_field=observed&interval=month'
    hist_month = requests.get(url).json()['results']['month']
    data = hist_month
    dates = list(data.keys())
    values = list(data.values())

    hist_obs = sns.barplot(x=dates, y = values, color = 'blue')
    hist_obs.set_xticklabels(hist_obs.get_xticklabels(), rotation = 90)
    return hist_obs
```

```python
get_hist_obs(264)
```

Mapa de color 

```python
def get_heatmap_obs(project_id: int):
    url  = f'{API_PATH}observations/?project_id={project_id}'
    per_page = 200
    total_number = requests.get(url).json()['total_results']

    for i in range(1,(total_number // per_page) + 2):
        results = requests.get(url).json()['results']
        for i in range(len(results)):
            if results[i]['quality_grade'] == 'research':
                longitud = results[i]['geojson']['coordinates'][0]
                latitud = results[i]['geojson']['coordinates'][1]
                m = folium.Map([np.mean(latitud), np.mean(longitud)], zoom_start=15)
                heat_data = ['latitud', 'longitud']
                HeatMap(heat_data).add_to(m)
            else:
                continue
    return m
```

```python
get_heatmap_obs(264)
```
